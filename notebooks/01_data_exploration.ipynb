```python
import pandas as pd

fake = pd.read_csv("/home/kinkini/Desktop/multilingual-fake-news-detector/datasets/raw/Fake.csv")
real = pd.read_csv("/home/kinkini/Desktop/multilingual-fake-news-detector/datasets/raw/True.csv")

print("Fake columns:")
print(fake.columns)

print("\nReal columns:")
print(real.columns)
```

```python
print(fake.shape)
print(real.shape)
```

```python
fake["label"] = 0
real["label"] = 1

df = pd.concat([fake, real])

df.head()
```

```python
df["label"].value_counts()
```

```python
import matplotlib.pyplot as plt
df["label"].value_counts().plot(
    kind="bar"
)

plt.title("Fake vs Real News")
plt.show()
```

```python
df["text_length"] = df["text"].astype(str).apply(len)

df["text_length"].describe()
```

```python
df["text_length"].hist(
    bins=50
)

plt.title("News Length Distribution")
plt.show()
```
