```python
import pandas as pd
from pathlib import Path
```

```python
DATA_PATH = Path(r"C:\Users\user\Documents\new fake news\NLP_Group_05\data")

fake_file = DATA_PATH / "Fake.csv"
true_file = DATA_PATH / "True.csv"
```

```python
fake_news = pd.read_csv(fake_file)
true_file = next(DATA_PATH.glob("True.csv"), None)
if true_file is None:
    true_file = next(DATA_PATH.glob("true.csv"), None)

if true_file is None:
    true_file = next(DATA_PATH.glob("Real.csv"), None)
    if true_file is None:
        true_file = next(DATA_PATH.glob("real.csv"), None)

    if true_file is None:
        raise FileNotFoundError(
            f"Could not find True.csv or Real.csv in {DATA_PATH}. "
            f"Available files: {[p.name for p in DATA_PATH.iterdir()]}"
        )

true_news = pd.read_csv(true_file)

print("Fake dataset shape:", fake_news.shape)
print("True dataset shape:", true_news.shape)
```

```python
display(fake_news.head())
display(true_news.head())
```

```python
fake_news["label"] = 0
true_news["label"] = 1
```

```python
combined_data = pd.concat(
    [fake_news, true_news],
    ignore_index=True
)

print("Total records:", combined_data.shape[0])
display(combined_data.head())
```

```python
print("Missing values:")
print(combined_data.isnull().sum())

print("\nClass distribution:")
print(combined_data["label"].value_counts())

print("\nDuplicate rows:")
print(combined_data.duplicated().sum())
```

```python
output_path = Path("../data/processed/combined_news.csv")

output_path.parent.mkdir(parents=True, exist_ok=True)

combined_data.to_csv(output_path, index=False)

print("Saved successfully to:", output_path)
```
