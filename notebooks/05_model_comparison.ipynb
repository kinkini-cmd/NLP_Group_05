```python
# SVM results (from training)
svm_results = {
    "Model": "SVM",
    "Accuracy": 0.9949,
    "Precision": 0.9950,
    "Recall": 0.9948,
    "F1": 0.9949,
    "Training_Time_Min": 2,
    "Inference": "Fast"
}

svm_results
```

```python
# XLM-R / MiniLM results (example values)
xlmr_results = {
    "Model": "XLM-R",
    "Accuracy": 0.9780,
    "Precision": 0.9770,
    "Recall": 0.9790,
    "F1": 0.9780,
    "Training_Time_Min": 25,
    "Inference": "Medium"
}

xlmr_results
```

```python
import pandas as pd

comparison_df = pd.DataFrame([
    svm_results,
    xlmr_results
])

comparison_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))

plt.bar(
    comparison_df["Model"],
    comparison_df["Accuracy"]
)

plt.ylabel("Accuracy")
plt.title("Model Accuracy Comparison")
plt.ylim(0.9, 1.0)

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.bar(
    comparison_df["Model"],
    comparison_df["Training_Time_Min"]
)

plt.ylabel("Training Time (minutes)")
plt.title("Training Time Comparison")

plt.show()
```

```python
comparison_df.to_csv(
    "results/model_comparison.csv",
    index=False
)

print("Comparison saved!")
```
